Consider the `CarPrice_Assignment.csv` data file. This data is publicly available on the Kaggle website, and has information on cars (characteristics related to car dimensions, engine and more). The goal is to use car information to predict the price of the car. **In Python**, answer the following:

### 1. (5 points) Using the `pandas` library, read the csv data file and create a data-frame called `car_price`.

```python
import pandas as pd

car_price = pd.read_csv('CarPrice_Assignment.csv')
car_price.head()
car_price.value_counts()
```

### 2. (5 points) Using the `wheelbase`, `enginesize`, `compressionratio`, `horsepower`, `peakrpm`, `citympg`, as the predictor variables, and `price` is the target variable. Define a 5-fold cross validation.

```python
from sklearn.model_selection import KFold

X = car_price[['wheelbase', 'enginesize', 'compressionratio', 'horsepower', 'peakrpm', 'citympg']]
y = car_price['price'].copy()

skf = KFold(n_splits=5, shuffle=True, random_state=42)
```

### 3. (10 points) Train and evaluate a MLP, with a single hidden layer with 10 neurons and the `ReLU` as the activation function. Also use `solver = 'adam'`, `max_iter = 100` and `batch_size = 256`, over the 5-fold cross validation. After that, use this model to predict on the test dataset. Report the average MSE of this model.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

mlp_model = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(10,),
        activation="relu",
        max_iter=100,
        solver="adam",
        batch_size=256,
        random_state=42

    )
)

# Running the cross-validation
cv_scores = cross_val_score(mlp_model, X, y, cv=skf, scoring='neg_root_mean_squared_error', n_jobs=-1)
print(f'The mean cross-validated MSE is {-cv_scores.mean()}')
```

### 4. (10 points) Train and evaluate a MLP, with two hidden layers: the first one with 10 neurons and the `ReLU` as the activation function, the second one with 8 neurons and `ReLU` as the activation function. Also use `solver = 'adam'`, `max_iter = 100` and `batch_size = 256`, over the 5-fold cross validation. After that, use this model to predict on the test dataset. Report the average MSE of this model.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

mlp_model = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(10,8),
        activation="relu",
        max_iter=100,
        solver="adam",
        batch_size=256,
        random_state=42

    )
)

# Running the cross-validation
cv_scores = cross_val_score(mlp_model, X, y, cv=skf, scoring='neg_root_mean_squared_error', n_jobs=-1)
print(f'The mean cross-validated MSE is {-cv_scores.mean()}')
```
